```python
import os
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras.applications import VGG16
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Flatten, Dense, Dropout
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau, ModelCheckpoint

# =====================================================
# DATASET PATH
# =====================================================
data_dir = "/content/CK+/CK+48"
save_path = "/content/ckplus_final_vgg16.keras"

# =====================================================
# DATA GENERATORS
# =====================================================
train_datagen = ImageDataGenerator(
    rescale=1./255,
    validation_split=0.2,
    rotation_range=15,
    width_shift_range=0.1,
    height_shift_range=0.1,
    zoom_range=0.1,
    horizontal_flip=True,
    fill_mode="nearest"
)

val_datagen = ImageDataGenerator(
    rescale=1./255,
    validation_split=0.2
)

train_gen = train_datagen.flow_from_directory(
    directory=data_dir,
    target_size=(224,224),
    batch_size=8,
    class_mode="categorical",
    subset="training",
    shuffle=True
)

val_gen = val_datagen.flow_from_directory(
    directory=data_dir,
    target_size=(224,224),
    batch_size=8,
    class_mode="categorical",
    subset="validation",
    shuffle=False
)

print("Classes:", train_gen.class_indices)

# =====================================================
# LOAD VGG16
# =====================================================
base_model = VGG16(
    weights="imagenet",
    include_top=False,
    input_shape=(224,224,3)
)

# Freeze all layers
for layer in base_model.layers:
    layer.trainable = False

# =====================================================
# BUILD MODEL
# =====================================================
x = Flatten()(base_model.output)
x = Dense(512, activation="relu")(x)
x = Dropout(0.5)(x)
x = Dense(256, activation="relu")(x)
x = Dropout(0.5)(x)
output = Dense(train_gen.num_classes, activation="softmax")(x)

model = Model(inputs=base_model.input, outputs=output)

model.summary()

# =====================================================
# CALLBACKS
# =====================================================
early_stop = EarlyStopping(
    monitor="val_accuracy",
    patience=7,
    restore_best_weights=True,
    verbose=1
)

reduce_lr = ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.5,
    patience=3,
    verbose=1
)

checkpoint = ModelCheckpoint(
    save_path,
    monitor="val_accuracy",
    save_best_only=True,
    verbose=1
)

# =====================================================
# STAGE 1
# =====================================================
model.compile(
    optimizer=Adam(learning_rate=1e-4),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

print("\n========== Stage 1 ==========\n")

history1 = model.fit(
    train_gen,
    validation_data=val_gen,
    epochs=25,
    callbacks=[early_stop, reduce_lr, checkpoint]
)

# =====================================================
# STAGE 2 (Fine-Tuning)
# =====================================================
for layer in base_model.layers[-4:]:
    layer.trainable = True

model.compile(
    optimizer=Adam(learning_rate=1e-5),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

print("\n========== Stage 2 ==========\n")

history2 = model.fit(
    train_gen,
    validation_data=val_gen,
    epochs=25,
    callbacks=[early_stop, reduce_lr, checkpoint]
)

# =====================================================
# EVALUATION
# =====================================================
loss, accuracy = model.evaluate(val_gen)

print(f"\nValidation Accuracy: {accuracy*100:.2f}%")
print(f"Validation Loss: {loss:.4f}")

# =====================================================
# SAVE MODEL
# =====================================================
model.save(save_path)

print("\nModel saved successfully!")
print(save_path)

# =====================================================
# PLOT RESULTS
# =====================================================
acc = history1.history["accuracy"] + history2.history["accuracy"]
val_acc = history1.history["val_accuracy"] + history2.history["val_accuracy"]

loss = history1.history["loss"] + history2.history["loss"]
val_loss = history1.history["val_loss"] + history2.history["val_loss"]

plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.plot(acc,label="Train Accuracy")
plt.plot(val_acc,label="Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.title("Accuracy")

plt.subplot(1,2,2)
plt.plot(loss,label="Train Loss")
plt.plot(val_loss,label="Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.title("Loss")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

from tensorflow.keras.preprocessing import image
from tensorflow.keras.models import load_model

# ==========================================
# LOAD MODEL
# ==========================================
model = load_model('/content/ckplus_final_vgg16.keras')

# ==========================================
# CLASS NAMES
# ==========================================
class_names = [
    'anger',
    'contempt',
    'disgust',
    'fear',
    'happy',
    'sadness',
    'surprise'
]

# ==========================================
# IMAGE PATH
# ==========================================
img_path = '/content/test_photo.jpg'   # Upload your image to Colab first

# ==========================================
# PREPROCESS IMAGE
# ==========================================
img = image.load_img(img_path, target_size=(224, 224))
img_array = image.img_to_array(img)
img_array = img_array / 255.0
img_array = np.expand_dims(img_array, axis=0)

# ==========================================
# PREDICT
# ==========================================
pred = model.predict(img_array, verbose=0)

predicted_index = np.argmax(pred)
predicted_emotion = class_names[predicted_index]
confidence = pred[0][predicted_index] * 100

# ==========================================
# DISPLAY IMAGE
# ==========================================
plt.figure(figsize=(5,5))
plt.imshow(image.load_img(img_path))
plt.axis("off")
plt.title(f"{predicted_emotion} ({confidence:.2f}%)")
plt.show()

# ==========================================
# PRINT RESULTS
# ==========================================
print(f"Predicted Emotion : {predicted_emotion}")
print(f"Confidence         : {confidence:.2f}%")

print("\nProbability of each emotion:")
for emotion, prob in zip(class_names, pred[0]):
    print(f"{emotion:<10}: {prob*100:.2f}%")
```
